## Ejercicio 4) Comprender el **Autoencoder**

En **clasificación**:

$$
x\longrightarrow y.
$$

En un **autoencoder**:

$$
x\longrightarrow z\longrightarrow\hat{x}.
$$

El **encoder** produce una **representación latente**:

$$
z=f_{\mathrm{enc}}(x),
$$

y el **decoder** reconstruye:

$$
\hat{x}=f_{\mathrm{dec}}(z).
$$

Usaremos un *bottleneck* de dimensión 64:

$$
784\longrightarrow64\longrightarrow784.
$$

Antes de continuar, responder brevemente:
- ¿qué representa $x$?

*x representa a la imágen de enetrada*
- ¿qué representa $z$?

*z representa a la transformación que se hace de la imágen x, la representación latente*
- ¿qué representa $\hat{x}$?

*es la imágen reconstruida*
- ¿por qué 64 constituye un *bottleneck* respecto de 784?

*Porque inicialmente tenemos una imágen de 784 pixeles y el cuello de botella viene de que a cada imágen la transformamos en imágenes de 64 pixeles sólo con información útil*.

## Ejercicio 5) Definir el autoencoder

Queremos un modelo completamente conectado con la estructura:

$$
(B,1,28,28)
\longrightarrow
(B,784)
\longrightarrow
(B,64)
\longrightarrow
(B,784)
\longrightarrow
(B,1,28,28).
$$

### Dropout opcional

El encoder incluirá `Dropout` con probabilidad configurable `p_dropout`.

**Ayudas:** completar la clase `Autoencoder`.

El **encoder** debe realizar conceptualmente:

```text
imagen → vector de 784 → capa lineal → ReLU → Dropout → código latente
```

El **decoder** debe realizar:

```text
código latente → capa lineal → Sigmoid → vector de 784
```

Finalmente, `forward()` debe recuperar la forma de imagen `(B,1,28,28)`.

**Ayudas:**
- utilizar `nn.Sequential`;
- utilizar `nn.Flatten`;
- recordar que `latent_dim` es un parámetro del constructor;
- la salida `Sigmoid` es coherente con píxeles en `[0,1]`.

````python
class Autoencoder(nn.Module):
    def __init__(self, latent_dim=64, p_dropout=0.2):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(28 * 28, latent_dim),
            nn.ReLU(),
            nn.Dropout(p_dropout)
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 28 * 28),
            nn.Sigmoid()
        )

    def forward(self, x):
        z = self.encoder(x)
        x_hat = self.decoder(z)

        # COMPLETAR:
        # Recuperar la forma de las imágenes de Fashion-MNIST.
        x_hat = x_hat.view(x.size(0), 1, 28, 28)

        return x_hat
````

## Ejercicio 8) Implementar una época de entrenamiento

La función debe recorrer todos los *batches* de *train* y actualizar los parámetros del modelo.

El `DataLoader` devuelve `images, labels`, pero las etiquetas no son necesarias.

Orden conceptual:

```text
activar modo entrenamiento
        ↓
recorrer batches
        ↓
mover imágenes al device
        ↓
borrar gradientes anteriores
        ↓
forward
        ↓
calcular MSE
        ↓
backpropagation
        ↓
actualizar parámetros
```

**Tarea:** completar la función sin calcular métricas adicionales dentro de ella.

````python
def train_one_epoch(model, data_loader, loss_fn, optimizer, device):
    model.train()

    for images, _ in data_loader:
        # COMPLETAR: mover imágenes al dispositivo.
        images = images.to(device)

        # COMPLETAR: borrar gradientes acumulados.
        optimizer.zero_grad()

        # COMPLETAR: forward.
        reconstructions = model(images)

        # COMPLETAR: calcular la pérdida.
        loss = loss_fn(reconstructions, images)

        # COMPLETAR: calcular gradientes.
        loss.backward()

        # COMPLETAR: actualizar parámetros.
        optimizer.step()
````

## Ejercicio 9) Implementar la evaluación

La evaluación **no debe modificar los parámetros**.

La función debe:
1. activar modo evaluación;
2. recorrer el `DataLoader`;
3. mover las imágenes al dispositivo;
4. obtener reconstrucciones;
5. calcular MSE;
6. acumular correctamente la pérdida;
7. devolver la pérdida promedio.

Para esta guía se puede decorar la función con `@torch.no_grad()`.

Como `nn.MSELoss()` devuelve un promedio del batch, usar el tamaño del batch al acumular:

```text
pérdida media del batch × cantidad de ejemplos del batch
```

**Tarea:** completar la función.

````python
@torch.no_grad()
def evaluate(model, data_loader, loss_fn, device):
    # COMPLETAR: activar modo evaluación.
    model.eval()

    total_loss = 0.0
    total_examples = 0

    for images, _ in data_loader:
        # COMPLETAR
        images = images.to(device)

        # COMPLETAR
        reconstructions = model(images)

        # COMPLETAR
        loss = loss_fn(reconstructions, images)

        # COMPLETAR
        batch_size = images.size(0)

        # COMPLETAR acumuladores.
        total_loss += loss.item() * batch_size
        total_examples += batch_size

    # COMPLETAR
    mean_loss = total_loss / total_examples

    return mean_loss
````

### 10.1) Reconstrucciones antes del entrenamiento

Antes de entrenar, los parámetros del modelo tienen valores iniciales aleatorios. Queremos observar qué reconstruye la red en ese estado.

Usar *validation*, no *test*.

**Tarea:** completar una función que:
1. activar modo evaluación;
2. seleccionar un *batch*;
3. conservar las primeras `n` imágenes;
4. obtener sus reconstrucciones;
5. llevar originales y reconstrucciones a CPU;
6. dibujar dos filas: originales arriba y reconstrucciones abajo.

No es necesario usar las etiquetas para calcular la reconstrucción; se pueden utilizarl sólo como títulos si así lo deseamos.

````python
@torch.no_grad()
def show_untrained_reconstructions(model, data_loader, device, n=6):
    # COMPLETAR
    model.eval()

    # COMPLETAR: tomar un batch.
    images, labels = next(iter(data_loader))

    # COMPLETAR: conservar n ejemplos y mover imágenes al device.
    images = images[:6].to(device)
    labels = labels[:6]

    # COMPLETAR: reconstrucciones.
    reconstructions = model(images)

    # COMPLETAR: llevar tensores necesarios a CPU.
    images = images.cpu()
    reconstructions = reconstructions.cpu()

    # COMPLETAR: crear figura 2 × n.
    fig, axes = plt.subplots(2, n, figsize=(3* n, 5))

    for i in range(n):
        # COMPLETAR fila de originales.
        axes[0, i].imshow(images[i].squeeze())

        # COMPLETAR fila de reconstrucciones.
        axes[1, i].imshow(reconstructions[i].squeeze())

    axes[0, 0].set_ylabel("Original")
    axes[1, 0].set_ylabel("Sin entrenar")

    fig.subplots_adjust(wspace=0.08, hspace=0.08)
    plt.tight_layout()
    plt.show()


# Cuando se complete la función, descomentar:
show_untrained_reconstructions(model, valid_loader, device)
````

En cada época debemos:

4. guardar ambas pérdidas;

Esto permite comparar train y validation en el mismo estado del modelo.

````python
EPOCHS = 10

history = {
    "train_loss": [],
    "valid_loss": []
}

for epoch in range(1, EPOCHS + 1):
    # COMPLETAR: entrenar una época.
    train_one_epoch(
        model,
        train_loader,
        loss_fn,
        optimizer,
        device
    )

    # COMPLETAR: evaluar train.
    train_loss = evaluate(
        model,
        train_loader,
        loss_fn,
        device
    )

    # COMPLETAR: evaluar validation.
    valid_loss = evaluate(
        model,
        valid_loader,
        loss_fn,
        device
    )

    # COMPLETAR: guardar pérdidas.
    history["train_loss"].append(train_loss)
    history["valid_loss"].append(valid_loss)

    # COMPLETAR: imprimir un resumen de la época.}
    print("-"*20)
    print(f"ÉPOCA N°{epoch}:")
    print(f"\t * Pérdida  de ENTRENAMIENTO obtenida: {round(train_loss, 4)}")
    print(f"\t * Pérdida de VALIDACIÓN obtenida: {round(valid_loss, 4)}")
````

## Ejercicio 15) Visualizar reconstrucciones sobre *test*

La función de pérdida (función error) MSE resume el error en un número, pero también debemos inspeccionar visualmente las reconstrucciones.

**Tarea:** completar una función que muestre:
- primera fila: imágenes originales;
- segunda fila: reconstrucciones del modelo entrenado.

Ahora sí corresponde utilizar `test_loader`, porque la configuración ya fue seleccionada.

````python
@torch.no_grad()
def show_reconstructions(model, data_loader, device, n=6):
    model.eval()

    # COMPLETAR: tomar un batch y seleccionar n imágenes.
    images, _ = next(iter(data_loader))
    images = images[:n].to(device)

    # COMPLETAR: reconstruir.
    reconstructions = model(images)

    # COMPLETAR: llevar a CPU.
    images = images.cpu()
    reconstructions = reconstructions.cpu()

    # COMPLETAR: figura 2 × n.
    n_mostrar = len(images)
    fig, axes = plt.subplots(
        2, n_mostrar, figsize=(1.4 * n_mostrar + 0.6, 3), squeeze=False
    )

    for i in range(n_mostrar):
        axes[0, i].imshow(images[i].squeeze(0), cmap="gray", vmin=0, vmax=1)
        axes[1, i].imshow(reconstructions[i].squeeze(0), cmap="gray", vmin=0, vmax=1)

        for ax in axes[:, i]:
            ax.set_xticks([])
            ax.set_yticks([])

    axes[0, 0].set_ylabel("Original")
    axes[1, 0].set_ylabel("Reconstrucción")

    fig.subplots_adjust(wspace=0.08, hspace=0.08)
    plt.show()

# Cuando se complete la función, descomentar:
show_reconstructions(model, test_loader, device)
````

## Ejercicio 17) Preguntas conceptuales

Responder.

1. ¿Por qué las etiquetas de Fashion-MNIST no son necesarias para entrenar el autoencoder?

> *Las etiquetas de Fashion-MNIST no son necesarias para entrenar porque no es el objetivo que buscamos el clasificar las prendas, sino reconstruirlas.*

2. ¿Qué representa el código latente?

> *El código latente es una representación interna que nos sirve para reconstruir la prenda.*

3. ¿Qué significa el bottleneck?

> *BottleNeck por su traducción directa significa cuello de botella, en este caso aparece en tomar una imágen de cierta cantidad de pixeles, y reducirla a menos pixeles tomando sólo información útil de la imágen.*


4. ¿Por qué usamos `MSELoss`?

> *Porque necesitamos medir una diferencia numérica entre la imagen original y la
reconstrucción. MSE promedia los errores cuadráticos de los píxeles. Un valor menor indica que, en promedio, los píxeles reconstruidos están más cerca de los originales.*


5. ¿Qué hace `optimizer.zero_grad()`?

> *Le indica al optimizador que no queremos que recalcule el gradiente, pues PyTorch por cada paso acumula gradientes.*


6. ¿Qué hace `loss.backward()`?

> *Realiza un cálculo de la pérdida para el back propagation, es decir calcula los gradientes pero no modifica los parámetros.*


7. ¿Qué hace `optimizer.step()`?

> *Realiza la actualización de los gradientes calculados, cambia pesos y bias.*


8. ¿Por qué volvemos a evaluar train después de cada época?

> *Porque queremos que el error entre cada época entre mini batch a mini batch se achique respecto al validation*.


9. ¿Qué diferencia existe entre `model.train()` y `model.eval()`?

> *La diferencia entre `model.train()` y `model.eval()` es que `model.train()` le indica al modelo que utilice los parámetros seteados para el entrenamiento, en este caso activa el Dropout, mientras que `model.eval()` le indica al modelo que vamos a evaluar, desactivando el Dropout*.


10. ¿Qué hace `Dropout(p)` durante `model.train()`?

> *Lo que hace `Dropout(p)` durante el entrenamiento es descartar el re-cálculo de algunos bias en el modelo, las activaciones que sobreviven se reescalan para mantener aproximadamente su valor esperado. Esto introduce ruido de regularización durante el entrenamiento.*

11. ¿Qué ocurre con dropout cuando ejecutamos `model.eval()`?

> *Se desactiva. La capa Dropout se comporta como identidad. De esta manera la inferencia es determinista para una entrada fija y utiliza todas las activaciones.*

12. ¿Qué significa elegir `DROPOUT_P = 0.0`?

> *Significa que con una probabilidad del 0.0 se seleccionarán capas de activación para realizar el cálculo de los gradientes, o dicho de otro modo, que no se apagarán capas para el cálculo del gradiente.*


13. ¿Por qué no colocamos dropout en la capa final que reconstruye los píxeles?

> *Porque la capa final representa directamente los píxeles reconstruidos. Anular
aleatoriamente componentes de esa salida introduciría ruido en los valores que estamos
comparando con la imagen objetivo. En esta guía Dropout actúa sobre la representación
interna.*


14. ¿Por qué SGD y Adam usan learning rates diferentes?

> *Porque los algoritmos transforman los gradientes de forma distinta. Adam adapta la magnitud de sus actualizaciones mediante estadísticas de los gradientes, mientras
que SGD básico usa directamente el gradiente multiplicado por el learning rate. Por eso sus escalas habituales de η no tienen que coincidir.*

````python
def reconstruccion(latent_dim, dropout_p, epochs, optimizer_name, learning_rate):
    torch.manual_seed(SEED)

    # COMPLETAR: crear el modelo y moverlo al device.
    model = Autoencoder(latent_dim=latent_dim, p_dropout=dropout_p).to(device)


    if optimizer_name == "Adam":
        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=learning_rate
        )
    elif optimizer_name == "SGD":
        optimizer = torch.optim.SGD(
            model.parameters(),
            lr=learning_rate
        )
    else:
        raise AttributeError("Error en el nombre del optimizador")

    print("Optimizador:", OPTIMIZER_NAME)

    history = {
        "train_loss": [],
        "valid_loss": []
    }

    for epoch in range(1, epochs + 1):
        # Entrenar una época.
        train_one_epoch(
            model,
            train_loader,
            loss_fn,
            optimizer,
            device
        )

        train_loss = evaluate(
            model,
            train_loader,
            loss_fn,
            device
        )

        valid_loss = evaluate(
            model,
            valid_loader,
            loss_fn,
            device
        )

        # guardar pérdidas.
        history["train_loss"].append(train_loss)
        history["valid_loss"].append(valid_loss)

        # imprimir un resumen de la época.
        print("-"*20)
        print(f"ÉPOCA N°{epoch} | P.Entrenamiento :  {round(train_loss, 4)} |  P.Validación: {round(valid_loss, 4)}")

    train = history["train_loss"]
    val = history["valid_loss"]

    epochs = range(1, epochs+1)

    plt.figure(figsize=(7, 4))

    # curvas de train y validation.
    plt.plot(epochs, train, marker="o", label="train")
    plt.plot(epochs, val, marker="o", label="validation")

    # etiquetas, título, leyenda y grilla.
    plt.xlabel("Épocas")
    plt.ylabel("MSE")
    plt.title(f"Autoencoder - {OPTIMIZER_NAME}")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

    return model, history
````
